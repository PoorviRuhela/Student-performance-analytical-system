```python
import pandas as pd
import numpy as np
df=pd.read_csv('/content/student_performance_20_students (1).csv')
print('Datasets load successfully')
df
```

```python
print('-------------------Datasets Informations -----------------')
print('1. Datasets shape')
print(df.shape)
print('2. Datasets columns information')
print(df.columns)
print('3. Datasets descriptions')
print(df.info())
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nStudent ID Duplicates:", df["Student_ID"].duplicated().sum())

print("\nInvalid Marks:")
marks_columns = ["Python_Marks", "NumPy_Marks", "Pandas_Marks"]

for col in marks_columns:
    print(col, ":", ((df[col] < 0) | (df[col] > 100)).sum())

print("\nInvalid Attendance Values:",
      ((df["Attendance_Percent"] < 0) |
       (df["Attendance_Percent"] > 100)).sum())
```

```python
df["Percentage"] = (
    df[marks_columns].mean(axis=1)
).round(2)

display(df[["Student_ID", "Name", "Percentage"]])
```

```python
def calculate_grade(percentage):
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 40:
        return "D"
    else:
        return "F"

df["Grade"] = df["Percentage"].apply(calculate_grade)

display(df[["Student_ID", "Name", "Percentage", "Grade"]])
```

```python
df["Result"] = np.where(
    (df[marks_columns] >= 33).all(axis=1),
    "Pass",
    "Fail"
)

display(df[[
    "Student_ID", "Name",
    "Python_Marks", "NumPy_Marks", "Pandas_Marks",
    "Percentage", "Grade", "Result"
]])

print("Total Passed:", (df["Result"] == "Pass").sum())
print("Total Failed:", (df["Result"] == "Fail").sum())
```

```python
print("CLASS PERFORMANCE SUMMARY")
print("-------------------------")

print("Total Students:", len(df))
print("Class Average Percentage:", round(df["Percentage"].mean(), 2))
print("Highest Percentage:", df["Percentage"].max())
print("Lowest Percentage:", df["Percentage"].min())
print("Passed Students:", (df["Result"] == "Pass").sum())
print("Failed Students:", (df["Result"] == "Fail").sum())
```

```python
subject_averages = df[marks_columns].mean().round(2)

print("Subject-wise Average Marks:")
display(subject_averages.to_frame(name="Average_Marks"))

best_subject = subject_averages.idxmax()
lowest_subject = subject_averages.idxmin()

print("Best Performing Subject:", best_subject)
print("Lowest Performing Subject:", lowest_subject)
```

```python
top_students = df.sort_values(
    by="Percentage",
    ascending=False
).head(3)

print("TOP 3 STUDENTS")
display(top_students[[
    "Student_ID", "Name", "Percentage", "Grade", "Result"
]])
```

```python
low_attendance = df[df["Attendance_Percent"] < 75]

print("Students with Attendance Below 75%:")
display(low_attendance[[
    "Student_ID", "Name", "Attendance_Percent"
]])

print("Low Attendance Students:", len(low_attendance))
print(
    "Class Average Attendance:",
    round(df["Attendance_Percent"].mean(), 2), "%"
)
```
